2. Extract and display the middle section of the image (150 pixels).

  • Extract a 150 pixel section from the center of the image using NumPy array slicing.
  
  • Display this cropped image using matplotlib.

```python
# Convert the image to a NumPy array
image_array = np.array(images)

# Get the dimensions of the image
height, width = image_array.shape[:2]

# Calculate the starting and ending indices for the center 150x150 region
start_x = (width // 2) - 75  # Center horizontally
start_y = (height // 2) - 75  # Center vertically
end_x = start_x + 150
end_y = start_y + 150

# Extract the middle 150x150 section using array slicing
middle_section = image_array[start_y:end_y, start_x:end_x]

# Display the cropped image using matplotlib
plt.imshow(middle_section)
plt.axis('off')  # Hide the axes
plt.title('Middle 150x150 Pixels')
plt.show()
```

3. Apply a simple threshold to the image (e.g., set all pixel values below 100 to 0).

  • Apply a threshold to the grayscale image: set all pixel values below 100 to 0, and all values above 100 to 255 (creating a binary image).

  • Display the resulting binary image.

```python
# Convert the image to grayscale (if it's not already grayscale)
gray_image = images.convert('L')

# Convert the grayscale image to a NumPy array
gray_array = np.array(gray_image)

# Apply the threshold: set all pixel values below 100 to 0, and all values >= 100 to 255
threshold_value = 100
binary_array = np.where(gray_array < threshold_value, 0, 255)

# Convert the binary array back to an image (optional, for saving or further processing)
binary_image = Image.fromarray(binary_array.astype(np.uint8))

# Display the binary image using matplotlib
plt.imshow(binary_image, cmap='gray')
plt.axis('off')  # Hide the axes
plt.title('Binary Image (Threshold = 100)')
plt.show()
```

5. Convert the grayscale image to an RGB image.

  • Convert the grayscale image into an RGB image where the grayscale values are replicated across all three channels (R, G, and B).

  • Display the converted RGB image using matplotlib.

```python
# Convert the image to grayscale (if it's not already grayscale)
gray_image = images.convert('L')

# Convert the grayscale image to a NumPy array
gray_array = np.array(gray_image)

# Expand the grayscale array into a 3-channel RGB array
rgb_array = np.stack((gray_array,) * 3, axis=-1)

# Convert the RGB array back to an image (optional, for saving or further processing)
rgb_image = Image.fromarray(rgb_array.astype(np.uint8))

# Display the converted RGB image using matplotlib
plt.imshow(rgb_image)
plt.axis('off')  # Hide the axes
plt.title('Grayscale to RGB Image')
plt.show()
```

Exercise 3: Image Compression and Decompression using PCA.
1. Load and Prepare Data:

  • Fetch an image of you choice.{If colour convert to grayscale}

```python
# Load the Image
img = Image.open("/content/drive/MyDrive/Artificial Intelligence/Week1/flowers.jpg")

# Convert the image to grayscale (if it's not already grayscale)
gray_image = img.convert('L')
display(gray_image)
```

• Center the dataset - Standaridze the Data.


```python
# Convert to Numpy array
gray_array = np.array(gray_image)

# Get image dimensions
height , width = gray_array.shape

# Copy the image data
data = gray_image.copy()

#compute the mean of each column
mean = np.mean(data, axis=0)

#substract mean to center the data
standardized_data = data-mean
print("Standardized Image Shape:",standardized_data.shape)

# Visualize the standardized image
plt.imshow(gray_array, cmap='gray')
plt.axis('off')
plt.title('Standardized Grayscale Image')
plt.show()
```


• Calculate the covaraince matrix of the Standaridze data.

```python
# Calculate the covariance matrix
cov_matrix = np.cov(standardized_data, rowvar=False)

# Display covariance matrix shape
print("Covariance Matrix:", cov_matrix)
```

2. Eigen Decomposition and Identifying Principal Components:

  • Compute Eigen Values and Eigen Vectors.

```python
# Compute Eigenvalues and Eigenvectors
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)
```

• Sort the eigenvalues in descending order and choose the top k eigenvectors corresponding to
the highest eigenvalues.

```python
# Sort eigenvalues and corresponding eigenvectors in descending order
sorted_indices = np . argsort ( eigenvalues ) [:: -1]
eigenvalues = eigenvalues [ sorted_indices ]
eigenvectors = eigenvectors [: , sorted_indices ]
```

Identify the Principal Components with the help of cumulative Sum plot.

```python
# Compute cumulative sum of eigenvalues
cumulative_variance = eigenvalues / np.sum(eigenvalues)

# Plot cumulative sum to identify principal components
plt.plot(np.cumsum(cumulative_variance))
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Cumulative Sum of Eigenvalues')
plt.grid()
plt.show()
```

3. Reconstruction and Experiment:

  • Reconstruction: Transform the original data by multiplying it with the selected eigenvectors (PCs) to obtain a lower-dimensional representation.
  Reconstruction and Experiment:

  • Experiments: Pick Four different combination of principal components with various explained variance value and compare the result.

  • Display the Results and Evaluate.

```python
# Select PC values (based on explained variance)
num_components_list = [10, 30, 100, 200]
reconstructed_images = []

for k in num_components_list:
    # Select top k eigenvectors
    selected_eigenvectors = eigenvectors[:, :k]

    # Transform and reconstruct image
    transformed_data = np.dot(standardized_data, selected_eigenvectors)
    reconstructed_data = np.dot(transformed_data, selected_eigenvectors.T) + mean
    reconstructed_images.append(reconstructed_data)

# Plot the original and reconstructed images
plt.figure(figsize=(12, 8))

plt.subplot(1, 5, 1)
plt.imshow(data, cmap='gray')
plt.title("Original Image")
plt.axis('off')

for i, (k, img) in enumerate(zip(num_components_list, reconstructed_images)):
    plt.subplot(1, 5, i + 2)
    plt.imshow(img, cmap='gray')
    plt.title(f"PCs: {k}")
    plt.axis('off')
plt.tight_layout()
plt.show()
```
